# **Marketing Campaign Analysis**

```python
import pandas as pd
import kagglehub


path = kagglehub.dataset_download("manishabhatt22/marketing-campaign-performance-dataset")
print("Dataset path:", path)

#importing data
df = pd.read_csv(f"{path}/marketing_campaign_dataset.csv")
print("Data Sample:")
display(df.head())

#data types check
print("\nData Information:")
df.info()
```

## **Data Cleaning: Analysing individual columns & fixing data types**

**Acquisition cost**

```python
df["Acquisition_Cost"] = (
    df["Acquisition_Cost"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["Acquisition_Cost"].value_counts()
```

**Duration**

```python
df["Duration"] = df["Duration"].str.replace(" days", "", regex=False).astype(int)
df["Duration"].value_counts()
```

**Date**

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Date"].head()
```

# **Exploratory Data Analysis**

**ROI efficiency by Channel**

```python
roi_df = df.groupby("Channel_Used")["ROI"].mean().reset_index()
roi_df = roi_df.sort_values(by="ROI", ascending=False).round(2)
roi_df
```

**Conversion Rate by Campaign Type**

```python
conversion_df = df.groupby("Campaign_Type")["Conversion_Rate"].mean().reset_index()
conversion_df = conversion_df.sort_values(by="Conversion_Rate", ascending=False)
conversion_df
```

**ROI VS Conversion**

```python
conversion_roi_df = (
    df.groupby("Campaign_Type")
      .agg(
          Conversion_Rate=("Conversion_Rate", "mean"),
          ROI=("ROI", "mean")
      )
      .reset_index()
)

conversion_roi_df
```
